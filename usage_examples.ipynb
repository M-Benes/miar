# Jupyter notebook with simple example usage of the MIAR package for calculating interestingness measures for rules from CleverMiner 4ft-miner

## Importing CleverMiner, MIAR and pandas

```python
from cleverminer import cleverminer
from miar.measures import *
import pandas as pd
```

## Dataset preparation and exploration

```python
df=pd.read_csv('AS.csv')
df=df[['Food and drink', 'Seat comfort', 'On-board service', 'Leg room service', 'Baggage handling','Checkin service', 'Inflight service', 'Cleanliness', 'satisfaction']]
print(df.isnull().sum())
print(df)
```

## Setting up CleverMiner 

```python
clm = cleverminer(df=df,proc='4ftMiner',
                  quantifiers= {'Base':4000, 'conf':0.8},
                  ante ={
                      'attributes':[
                    {'name': 'Food and drink', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                    {'name': 'Seat comfort', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                    {'name': 'On-board service', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                    {'name': 'Leg room service', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                    {'name': 'Baggage handling', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                    {'name': 'Checkin service', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                    {'name': 'Inflight service', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                        {'name': 'Cleanliness', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                      ], 'minlen':1, 'maxlen':5, 'type':'con'},
                  succ ={
                      'attributes':[
                        {'name': 'satisfaction', 'type': 'seq', 'minlen': 1, 'maxlen': 1},
                      ], 'minlen':1, 'maxlen':1, 'type':'con'}
                  )
```

## Setting up a simple loop to iterate over results from cleverminer

```python
for i in (range(clm.get_rulecount())):
        rule_id = i+1
        arr = clm.get_fourfold(rule_id)
        criterion = zhang(arr)
        print(f"Rule {rule_id}: {clm.get_ruletext(rule_id)} Zhang:{criterion}")
```

## Combining multiple criteria to filter rules

```python
for i in (range(clm.get_rulecount())):
        rule_id = i+1
        arr = clm.get_fourfold(rule_id)
        criterion = zhang(arr)
        criterion2 = yule_y(arr)
        if (criterion > 0.9) and (criterion2>0.7):
            print(f"Rule {rule_id}: {clm.get_ruletext(rule_id)} Zhang:{criterion} YuleY:{criterion2}")


```
